#2. Removing or Capping Outliers

Use the IQR (Interquartile Range) method to cap outliers instead of deleting them

###Delivery_Time

```python
fdd_df['Delivery_Time_Min'].describe()
```

```python
fdd_df['Delivery_Time_Min'].plot(kind='box')
```

```python
Q1 = fdd_df["Delivery_Time_Min"].quantile(0.25)
Q3 = fdd_df["Delivery_Time_Min"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
outliers = fdd_df[
    (fdd_df["Delivery_Time_Min"] < lower_limit) |
    (fdd_df["Delivery_Time_Min"] > upper_limit)
]

print("Number of Outliers:", len(outliers))
```

Delivery_Time_Min was examined using the IQR method. The calculated lower limit was -102.5 and the upper limit was 325.5. Since all values (20–300 minutes) fall within this range, no outliers were detected and no capping or removal was required.

###5.4 Delivery performance categories

```python
def delivery_performance(time):
    if time <= 30:
        return "Fast"
    elif time <= 60:
        return "Moderate"
    elif time <= 120:
        return "Slow"
    else:
        return "Very Slow"

fdd_df["Delivery_Performance"] = fdd_df["Delivery_Time_Min"].apply(delivery_performance)

fdd_df["Delivery_Performance"].value_counts()
```

```python
fdd_df["Delivery_Performance"] = fdd_df["Delivery_Time_Min"].apply(delivery_performance)

fdd_df["Delivery_Performance"].value_counts()
```

```python
bins = [0, 30, 60, 120, float("inf")]
labels = ["Fast", "Moderate", "Slow", "Very Slow"]

fdd_df["Delivery_Performance"] = pd.cut(
    fdd_df["Delivery_Time_Min"],
    bins=bins,
    labels=labels
)

fdd_df["Delivery_Performance"].value_counts()
```

```python
fdd_df[["Delivery_Time_Min", "Delivery_Performance"]].head(10)
```

Feature: Delivery Performance Categories

A new feature, Delivery_Performance, was derived from the delivery time to classify each order into four categories: Fast (≤30 minutes), Moderate (31–60 minutes), Slow (61–120 minutes), and Very Slow (>120 minutes). This categorization simplifies the evaluation of delivery efficiency and helps identify areas where operational improvements may be required.

11. Cancellation Rate by Restaurant

```python
total_orders = (
    fddc_df.groupby("Restaurant_Name")
    .size()
    .reset_index(name="Total_Orders")
)

cancelled_orders = (
    fddc_df[fddc_df["Order_Status"] == "Cancelled"]
    .groupby("Restaurant_Name")
    .size()
    .reset_index(name="Cancelled_Orders")
)

cancel_rate = total_orders.merge(
    cancelled_orders,
    on="Restaurant_Name",
    how="left"
)

cancel_rate["Cancelled_Orders"] = cancel_rate["Cancelled_Orders"].fillna(0)

cancel_rate["Cancellation_Rate (%)"] = (
    cancel_rate["Cancelled_Orders"] /
    cancel_rate["Total_Orders"] * 100
)

cancel_rate = cancel_rate.sort_values(
    "Cancellation_Rate (%)",
    ascending=False
)

cancel_rate.head(10)
```

```python
top_cancel = cancel_rate.head(10)

plt.figure(figsize=(10,5))

plt.bar(
    top_cancel["Restaurant_Name"],
    top_cancel["Cancellation_Rate (%)"]
)

plt.title("Top 10 Restaurants by Cancellation Rate")
plt.xlabel("Restaurant")
plt.ylabel("Cancellation Rate (%)")
plt.xticks(rotation=90)

plt.show()
```
